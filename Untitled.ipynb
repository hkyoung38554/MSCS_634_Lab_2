```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier, RadiusNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
## Step 1: Load and Prepare the Dataset

# This lab uses the Wine dataset from `sklearn`, which contains 13 chemical features and a 3-class target label representing different wine cultivars.
```

```python
wine = load_wine()
X = wine.data
y = wine.target

df = pd.DataFrame(X, columns=wine.feature_names)
df["class"] = y

print("Dataset shape (rows, cols):", df.shape)
print("\nFeature columns:")
print(wine.feature_names)

print("\nFirst 5 rows:")
display(df.head())

print("\nClass distribution (counts):")
print(df["class"].value_counts().sort_index())

print("\nClass distribution (percent):")
print((df["class"].value_counts(normalize=True).sort_index() * 100).round(2))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train size:", X_train.shape, " Test size:", X_test.shape)
```

```python
## Step 2: Implement K-Nearest Neighbors (KNN)

# We train and evaluate KNN models using k values: 1, 5, 11, 15, and 21.  
# Accuracy is computed on the 20% test set.
```

```python
k_values = [1, 5, 11, 15, 21]
knn_results = []

for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)
    knn_results.append({"k": k, "accuracy": acc})

knn_df = pd.DataFrame(knn_results)
knn_df
```

```python
## Step 3: Implement Radius Neighbors (RNN)

# We train and evaluate Radius Neighbors models using radius values: 350, 400, 450, 500, 550, and 600.  
# Accuracy is computed on the 20% test set.
```

```python
radius_values = [350, 400, 450, 500, 550, 600]
rnn_results = []

for r in radius_values:
    model = RadiusNeighborsClassifier(radius=r)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)
    rnn_results.append({"radius": r, "accuracy": acc})

rnn_df = pd.DataFrame(rnn_results)
rnn_df
```

```python
## Step 4: Visualize and Compare Results

# We plot accuracy trends for KNN (by k) and RNN (by radius), then compare model behavior.
```

```python
plt.figure()
plt.plot(knn_df["k"], knn_df["accuracy"], marker="o")
plt.title("KNN Accuracy vs k")
plt.xlabel("k (number of neighbors)")
plt.ylabel("Accuracy")
plt.xticks(knn_df["k"])
plt.ylim(0, 1.05)
plt.grid(True)
plt.show()
```

```python
plt.figure()
plt.plot(rnn_df["radius"], rnn_df["accuracy"], marker="o")
plt.title("RNN Accuracy vs Radius")
plt.xlabel("Radius")
plt.ylabel("Accuracy")
plt.xticks(rnn_df["radius"])
plt.ylim(0, 1.05)
plt.grid(True)
plt.show()
```

```python
## Results Summary (Test Accuracy)

# Using `random_state=42` and a stratified 80/20 split, the following accuracies were observed:

# **KNN**
# - k=1  → 0.7778  
# - k=5  → 0.8056  
# - k=11 → 0.8056  
# - k=15 → 0.8056  
# - k=21 → 0.8056  

# **RNN**
# - radius=350 → 0.7222  
# - radius=400 → 0.6944  
# - radius=450 → 0.6944  
# - radius=500 → 0.6944  
# - radius=550 → 0.6667  
# - radius=600 → 0.6667  
```

```python
best_knn_k = knn_df.sort_values("accuracy", ascending=False).iloc[0]["k"]
best_rnn_radius = rnn_df.sort_values("accuracy", ascending=False).iloc[0]["radius"]

best_knn = KNeighborsClassifier(n_neighbors=int(best_knn_k))
best_knn.fit(X_train, y_train)
knn_preds = best_knn.predict(X_test)

best_rnn = RadiusNeighborsClassifier(radius=float(best_rnn_radius))
best_rnn.fit(X_train, y_train)
rnn_preds = best_rnn.predict(X_test)

print("Best KNN k:", best_knn_k)
print("Confusion Matrix (KNN):\n", confusion_matrix(y_test, knn_preds))
print("\nClassification Report (KNN):\n", classification_report(y_test, knn_preds))

print("\nBest RNN radius:", best_rnn_radius)
print("Confusion Matrix (RNN):\n", confusion_matrix(y_test, rnn_preds))
print("\nClassification Report (RNN):\n", classification_report(y_test, rnn_preds))
```

```python
## Observations and Discussion

### Accuracy trends
# - **KNN** improved from k=1 to k=5 and then stayed flat (0.8056) for larger k values in this run.  
  This suggests that with unscaled raw features, increasing k beyond a small neighborhood did not provide additional benefit on the test split used.
- **RNN** achieved the best accuracy at **radius=350 (0.7222)** and gradually declined as radius increased to 600 (0.6667).  
  Increasing radius includes more neighbors in the vote, which can “blur” class boundaries and reduce accuracy.

### KNN vs RNN performance
- In these results, **KNN outperformed RNN** across the tested parameter ranges (best KNN 0.8056 vs best RNN 0.7222).
- KNN uses a **fixed number of neighbors**, making it more stable across regions of different sample density.
- RNN uses a **fixed distance radius**, so prediction behavior can vary more depending on how dense or sparse training points are around each test sample.

### When KNN might be preferable
- When you want consistent behavior everywhere (always use k neighbors).
- When dataset density varies, and you still want a stable vote size.

### When RNN might be preferable
- When a distance threshold is meaningful for the problem domain.
- When you want the model to adapt neighbor counts automatically (few neighbors in sparse areas, more in dense areas), as long as the radius is well-tuned.
```
